# DrillRheo Quickstart

Fits Bingham Plastic, Power Law, and Herschel-Bulkley models to a Fann viscometer dataset, ranks them by AICc, and generates figures + a report.

```python
from drillrheo.data_input import load_fann_data
from drillrheo.fitting import fit_all
from drillrheo.model_selection import compare_models, summarize
from drillrheo.report import generate_report, plot_rheogram

df = load_fann_data("sample_data.csv")
df
```

## Fit all models

```python
results = fit_all(df)
results.keys()
```

## Compare models statistically (AICc, not just R²)

```python
comparison = compare_models(results, df["shear_rate_1s"], df["shear_stress_pa"])
print(summarize(comparison))
comparison
```

## Plot the rheogram

```python
import matplotlib.pyplot as plt

plot_rheogram(df, results, title="Sample fluid")
plt.show()
```

## Generate the full report (figures + JSON/CSV)

```python
paths = generate_report(df, results, comparison, output_dir="quickstart_output", name="sample_fluid")
paths
```
